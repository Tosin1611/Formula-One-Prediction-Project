```python
constructors_renamed = constructors[['constructorId', 'name']].rename(columns={'name': 'constructor_name'})
circuits_renamed = circuits[['circuitId', 'name', 'location', 'country']].rename(columns={'name': 'circuit_name'})

merged = merged.merge(constructors_renamed, on='constructorId', how='left')
merged = merged.merge(circuits_renamed, on='circuitId', how='left')
```

```python
merged_sorted = merged.sort_values(['year', 'round'])
merged_sorted[['raceId', 'year', 'round']].drop_duplicates().head(20)
```

```python
merged_sorted = merged.sort_values(['year', 'round']).reset_index(drop=True)

merged_sorted['recent_form'] = (
    merged_sorted.groupby('driverId')['positionOrder']
    .transform(lambda x: x.shift(1).rolling(5).mean())
)

merged_sorted[['driverId', 'year', 'round', 'positionOrder', 'recent_form']].head(20)
```
